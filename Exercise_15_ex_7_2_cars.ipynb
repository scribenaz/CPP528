# Exercise 7-2: Prepare the Cars data

## Read the data

```python
import pandas as pd
```

```python
cars = pd.read_csv('cars.csv')
```

```python
# display the first five rows
```

```python
cars.head()
```

## Add and drop columns

```python
# display the unique values for the CarName column
```

```python
cars['CarName'].unique()
```

```python
# add a column for the brand by using a lambda expression
```

```python
cars['brand'] = cars['CarName'].apply(lambda x: x.split(' ')[0])
cars.head()
```

```python
# add a column for the model by using a user-defined function
```

```python
def get_model(row):
    words = row.CarName.split(' ')
    if len(words) > 1:
        return ' '.join(words[1:])
    else:
        return ''
cars['model'] = cars.apply(get_model, axis =1)
cars.head()
```

```python
# drop the CarName and car_ID columns
```

```python
cars = cars.drop(columns = ['CarName', 'car_ID'])
cars.head()
```

## Set an index and unstack the data

```python
# filter the cars data and store in a variable named cars_filtered
# only get rows for Volkswagon
# only get the columns named model, horsepower, carbody, and doornumber
```

```python
cars_filtered = cars[cars['brand'] == 'volkswagen']
cars_filtered = cars_filtered[['model', 'horsepower', 'carbody', 'doornumber']]
cars_filtered.head()
```

```python
# index the cars_filtered data by model, carbody, and doornumber
```

```python
cars_indexed = cars_filtered.set_index(['model', 'carbody', 'doornumber'])
cars_indexed
```

```python
# unstack the carbody column
```

```python
cars_indexed.unstack('carbody')
```
