## Second Pair

Looking into the data, there appears to be no accidents, yet there are collisions in the Collision_Type column as well as dispatches in the Dispatch_Priority column but in the Accident_Occurred column we just see 200000 and that it is non-null with an int64 data type. We can check the value of accidents by using the sum method. Which shows us 8231 to match the Collision_type and Dispatch_Priority values.

```python
total_accident_occurred = df['Accident_Occurred'].sum()
print(total_accident_occurred)
```

I created a variable to put into a dataframe the instances where an accident occurred is True, i printed it out which just showed me the Boolean Values of True or False.

```python
Accidents = df['Accident_Occurred']==1
print(Accidents)
```

By taking it one step further, I put df[Accidents] to see the dataframe I just created in the previous step and now I can see the rows where an accident occurred, which shows up as a value of 1 and at the bottom we can see that there are 8231 rows which match the number of Collision_Type and Dispatch_Priority values.

```python
df[Accidents]
```
