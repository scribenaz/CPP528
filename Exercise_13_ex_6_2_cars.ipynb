## Fix spelling and capitalization problems in the data

```python
cars['brand'] = cars.apply(lambda x: x.CarName.split(' ')[0], axis=1)
cars['name'] = cars.apply(lambda x: ' '.join(x.CarName.split(' ')[1:]), axis=1)
```

```python
# display each unique value in the brand column
```

```python
print(cars['brand'].unique())
```

```python
# fix any misspellings or inconsistent capitalization in the brand names
```

Fixing the capitalization of the cars brand name.

```python
cars['brand'] = cars['brand'].str.title()
print(cars['brand'].unique())
```

Still issues with BMW and VW.

```python
cars['brand'] = cars['brand'].replace('Vw', 'VW')
cars['brand'] = cars['brand'].replace('Bmw', 'BMW')
print(cars['brand'].unique())
```

Capitalization looks good, now for the misspellings.

```python
cars['brand'] = cars['brand'].replace('Mazdz', 'Mazda')
cars['brand'] = cars['brand'].replace('Toyouta', 'Toyota')
cars['brand'] = cars['brand'].replace('Porcshce', 'Porsche')
cars['brand'] = cars['brand'].replace('Vokswagen', 'Volkswagen')
print(cars['brand'].unique())
```

Misspellings are corrected and looks good.

```python
# store the fixed data in the CarName column
```

```python
cars['CarName']  = cars['brand'] + ' ' + cars['name']
cars.head(5)
```

The car brand names looks good, we were not asked to correct the car name column.

## Rename and drop columns

```python
# rename the CarName and the car_ID columns to brandname and carid
```

```python
cars.rename(columns={'CarName': 'brandname', 'car_ID': 'carid'}, inplace=True)
cars.head(5)
```

```python
# drop the carid and symbol columns
```

```python
cars.drop(['carid', 'symboling'], axis=1, inplace=True)
cars.head(5)
```
