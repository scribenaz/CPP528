# Exercise 8-1: Analyze the Forest Fires data

## Read the data

```python
import pandas as pd
```

```python
fires_by_month = pd.read_pickle('fires_by_month.pkl')
```

```python
# display the first five rows
```

```python
fires_by_month.head()
```

```python
fires_by_month.info()
```

## Group and aggregate the data

```python
# group the data by state and year
```

```python
fires_grouped = fires_by_month.groupby(['state', 'fire_year'], observed=True)
```

```python
# sum the grouped data and store DataFrame in fires_by_year variable
```

```python
fires_by_year = fires_grouped.sum()
fires_by_year
```

```python
# drop the fire_month column
```

```python
fires_by_year.drop('fire_month', axis=1, inplace=True)
fires_by_year
```

## Use pivot tables

```python
# select rows where the year is 2013 or later. Then, reset the index for the DataFrame that's returned and assign the DataFrame to a variable named fires_recent.
```

```python
fires_recent = fires_by_year.query('fire_year >= 2013').reset_index()
fires_recent.head()
```

```python
# use the pivot() method to pivot the data so the state values are used as row labels,
# fire_year values are used as column labels, and acres_burned values are used as data
```

```python
fires_recent.pivot(index='state', columns='fire_year', values='acres_burned')
```

```python
# use the pivot_table() method with the fires_by_month DataFrame to get the same result
# as the previous cell
```

```python
fires_by_month.query('fire_year >= 2013').pivot_table(index='state', columns='fire_year', values='acres_burned', aggfunc='sum')
```

## Work with bins

```python
# reset the index for the DataFrame named fires_by_year
```

```python
fires_by_year = fires_by_year.reset_index()
fires_by_year.head()
```

```python
# bin the data by decade
```

```python
fires_by_year['decade'] = pd.cut(fires_by_year.fire_year, bins =[1989, 1999, 2009, 2019], labels=['1990s', '2000s', '2010s'])

```

```python
# double-check the edge values
```

```python
fires_by_year.head(25)
```

```python
# drop the fire_year column and assign the DataFrame that's returned to a variable named fires_by_decade
```

```python
fires_by_decade = fires_by_year.drop('fire_year', axis=1)
fires_by_decade.head()
```

```python
# group by the state and decade columns and sum the data
```

```python
fires_by_decade = fires_by_decade.groupby(['state', 'decade'], observed=True).sum()
fires_by_decade
```
