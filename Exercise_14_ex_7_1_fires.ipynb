## Add and modify columns

```python
# add a column for the mean number of acres burned per day
```

```python
fires_by_month['acres_per_day'] = fires_by_month.dropna().acres_burned / fires_by_month.dropna().days_burning
fires_by_month['acres_per_day'] = fires_by_month['acres_per_day']
fires_by_month.head()
```

```python
# add a column that uses a lambda to calculate the mean number of acres burned per day
```

```python
fires_by_month['mean_acres_per_day'] = fires_by_month.apply(lambda x: x.acres_burned / x.days_burning if x.days_burning > 0 else 0, axis=1)
fires_by_month['mean_acres_per_day'] = fires_by_month['mean_acres_per_day']
fires_by_month.head()
```

```python
(fires_by_month['days_burning'] == 0).sum()
```

```python
fires_by_month[fires_by_month['days_burning'] == 0].head()
```

```python
# write a function to convert the fire_month column from an int value to a string value such as 'Jan', 'Feb', etc.
```

```python

def convert_fire_month(row):
  if row.fire_month == 1:
    return 'Jan'
  elif row.fire_month == 2:
    return 'Feb'
  elif row.fire_month == 3:
    return 'Mar'
  elif row.fire_month == 4:
    return 'Apr'
  elif row.fire_month == 5:
    return 'May'
  elif row.fire_month == 6:
    return 'Jun'
  elif row.fire_month == 7:
    return 'Jul'
  elif row.fire_month == 8:
    return 'Aug'
  elif row.fire_month == 9:
    return 'Sep'
  elif row.fire_month == 10:
    return 'Oct'
  elif row.fire_month == 11:
    return 'Nov'
  else:
    return 'Dec'





```

```python
# apply the function to the fire_month column
```

```python
fires_by_month['fire_month'] = fires_by_month.apply(convert_fire_month, axis=1)
fires_by_month.head()
```

```python
fires_by_month.head(5)
```

## Work with indexes

```python
# set an index on the state, fire_year, and fire_month columns
```

```python
fires_by_month = fires_by_month.set_index(['state', 'fire_year', 'fire_month'])
fires_by_month.head()
```

```python
# unstack the fire_month column and store the resulting DataFrame in a different variable
```

```python
fires_by_month_wide = fires_by_month.unstack('fire_month')
fires_by_month_wide.head()
```

```python
# reset the index for the fires_by_month DataFrame
```

```python
fires_by_month = fires_by_month.reset_index()
fires_by_month.head()
```

## Combine data

```python
# create new fire data
new_fire = pd.DataFrame(data=[['CA',2021,'June',1000,100,1,10,10]], columns=fires_by_month.columns)
new_fire.head()
```

```python
# add the new fire data to the end of the original DataFrame
```

```python
fires_by_month = pd.concat([fires_by_month, new_fire])
```
