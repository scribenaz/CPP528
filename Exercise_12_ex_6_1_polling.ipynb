# Exercise 6-1: Clean the Polling data

## Read the data

```python
import pandas as pd
```

```python
polls = pd.read_csv('president_polls_2016.csv')
```

## Examine the data

```python
# run the info() method on the data
```

Columns that have missing values: grade, sample_size, rawpoll_johnson, rawpoll_mcmullin, adpoll_johnson, adpoll_mcmullin, multiversions, and url.

```python
polls.info(verbose=True, memory_usage='deep', show_counts=True)
```

```python
# run the nunique() method on the data
```

Columns with only one value: cycle, branch, matchup, forecastdate, and multiversions

```python
polls.nunique()
```

## Drop columns and rows

```python
# drop all rows except where the column is "now-cast"
```

```python
polls = polls[polls.type == 'now-cast']
```

```python
# run the nunique() method again
```

```python
polls.nunique()
```

```python
polls.info()
```

```python
# drop columns with a single value
```

```python
polls = polls.drop(columns=['cycle', 'branch', 'type', 'matchup', 'forecastdate', 'multiversions', 'timestamp'])
```

```python
polls.nunique()
```

```python
# drop rows with "U.S." for the state
```

```python
polls = polls[polls.state != 'U.S.']
```

```python
polls.nunique()
```

Result: After running the commands to drop rows and columns we can see that our number of values the nunique() method produces has decreased by quite a bit. We are successfully cleaning up the data and narrowing down our focus while dropping information that may not actually be useful.

## Rename columns

```python
# rename all the rawpoll_* columns
```

```python
polls = polls.rename(columns={
    'rawpoll_clinton': 'clinton_pct',
    'rawpoll_trump': 'trump_pct',
    'rawpoll_johnson': 'johnson_pct',
    'rawpoll_mcmullin': 'mcmullin_pct'})
```

## Fix data types

```python
# run the info() method again
```

```python
polls.info()
```

```python
# create a list of columns that should be datetimes
```

These are the columns that should be using the datetime data type: startdate, enddate, createddate.

```python
date_columns = ['startdate', 'enddate', 'createddate']
```

```python
# convert the columns to datetime
```

```python
polls[date_columns] = polls[date_columns].apply(pd.to_datetime)
```

```python
polls.info()
```

```python
# convert the state and population columns to categories
```

```python
polls = polls.astype({'state': 'category', 'population': 'category'})
```

```python
polls.info()
```
